Build a Basic chatBot with LangGraph

```python
from typing_extensions import TypedDict
from typing import Annotated

from langgraph.graph import StateGraph,START,END
from langgraph.graph.message import add_messages

```

```python
class State(TypedDict):
    messages: Annotated[list, add_messages]

# OUTDENT THIS LINE so it is outside the class
graph_builder = StateGraph(State)
```

```python
import os 
from dotenv import load_dotenv
load_dotenv()

```

```python
import os
from langchain_openai import ChatOpenAI
from langchain.chat_models import init_chat_model
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
os.environ["GROQ_API_KEY"]= os.getenv("GROQ_API_KEY")

llm=ChatGroq(model="groq:qwen/qwen3-32b")



```

```python
llm
```

```python
def chatbot(state: State):
    return {"messages": [llm.invoke(state["messages"])]}
```

```python
graph_builder=StateGraph(State)
## Adding nodes
graph_builder.add_node("llmchatbot",chatbot)
##adding edges
graph_builder.add_edge(START,"llmchatbot")
graph_builder.add_edge("llmchatbot",END)
# compile the graph
graph=graph_builder.compile()

```

```python
from IPython.display import Image, display

try:
    display(Image(graph.get_graph().draw_mermaid_png()))
except Exception:
    # This requires additional system dependencies (like Graphviz/pygraphviz) 
    # and might be skipped if they aren't installed.
    pass
```

```python
graph.invoke({"messages":"Hi"})
```

```python
response =graph.invoke({"messages":"How to learn Agentic AI"})
response["messages"][-1].content
```

```python
# Pass the query as a list of message tuples, and use .values()
for event in graph.stream({"messages": [("user", "Is Agentic AI difficult to learn ?")]}):
    for value in event.values():
        print(value["messages"][-1].content)
```

```python
import os
from dotenv import load_dotenv
# Force-reload the .env file to pick up newly added keys
load_dotenv(override=True)

from langchain_tavily import TavilySearch

# Corrected parameter name: max_results (with an 's')
tool = TavilySearch(max_results=2)

res = tool.invoke("What is difference between langchain and LangGraph")
print(res)
```

```python
def multiply(a:int,b:int)->int:
    """Multiply a and b

    Args:
    a (int) : first int
    b (int) :second int

    Returns:
    int: output int
    """
```

```python
tools=[tool,multiply]
```

```python
llm_with_tool=llm.bind_tools(tools)
```

```python
llm_with_tool
```

```python
from IPython.display import Image, display
from langgraph.graph import StateGraph, START, END
from langgraph.prebuilt import ToolNode, tools_condition

# 1. Definition of Node (uses the State class defined in the previous cell)
def tool_calling_llm(state: State):
    return {"messages": [llm_with_tool.invoke(state["messages"])]}

# 2. Graph setup
builder = StateGraph(State)
builder.add_node("tool_calling_llm", tool_calling_llm)
builder.add_node("tools", ToolNode(tools)) # Make sure `tools` list is defined

# 3. Add edges
builder.add_edge(START, "tool_calling_llm")
builder.add_conditional_edges(
    "tool_calling_llm",
    tools_condition
)
builder.add_edge("tools", "tool_calling_llm") # Typically loops back to the LLM after tool execution

graph = builder.compile()

# 4. Draw graph
try:
    display(Image(graph.get_graph().draw_mermaid_png()))
except Exception:
    # Fallback to printing mermaid markdown if pygraphviz isn't installed
    print(graph.get_graph().draw_mermaid())
```
